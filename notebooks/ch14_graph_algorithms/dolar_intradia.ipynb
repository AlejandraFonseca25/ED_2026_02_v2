```python
import pandas as pd

df = pd.read_csv(
    "intradia.csv",
    sep=";",
    decimal=",",
    encoding="utf-8-sig",   # quita el BOM del encabezado
    parse_dates=["Fecha"],
)
df = df.rename(columns={"COP/USD": "precio"})
df = df.sort_values("Fecha").reset_index(drop=True)

df.head()
```

```python
df.info()
df["precio"].describe()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(df["Fecha"], df["precio"], linewidth=1)
ax.set_title("COP/USD intradía")
ax.set_xlabel("Hora")
ax.set_ylabel("Pesos por dólar")
ax.grid(alpha=0.3)
plt.show()
```
